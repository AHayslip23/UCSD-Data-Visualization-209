```python
allegations
```

```python
allegations["Allegation Month and Year Value"] = pd.to_datetime(allegations["Allegation Month and Year"])
```

```python
allegations["After Neighborhood Policing"] = np.where(
    (allegations["Allegation Month and Year Value"] >= "2015-06-01") &
    (allegations["Allegation Month and Year Value"] <= "2019-12-31"),
    True,
    False
)
```

```python
allegations["After Neighborhood Policing"].value_counts()
```

```python
Data20112019 = allegations[allegations["Year When Complaint Was Received"].isin([2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019])]
```

```python
stacks = (
    Data20112019.groupby(["Ethnicity of Compliant", "After Neighborhood Policing"])
    .size()
    .unstack(fill_value=0)
)
```

```python
stacks
```

```python
stack_data = stacks.loc[stacks.sum(axis=1).sort_values(ascending=False).index]
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


x = np.arange(len(stack_data.index))  
width = 0.4  

plt.figure(figsize=(10, 6))
colors = sns.color_palette("Set2", n_colors=2)

plt.bar(x - width/2, stack_data[False], width, label="Before (Jan 2011–May 2015)", color=colors[0])
plt.bar(x + width/2, stack_data[True],  width, label="After (Jun 2015–Dec 2019)", color=colors[1])

plt.title("Complaints by Ethnicity Before and After Neighborhood Policing (2011–2019)", fontsize=14)
plt.xlabel("Ethnicity of Complainant", fontsize=12)
plt.ylabel("Number of Complaints", fontsize=12)
plt.xticks(x, stack_data.index)
plt.legend(title="After Neighborhood Policing", loc="upper right")
plt.tight_layout()
plt.show()
```
